```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sentence_transformers import SentenceTransformer
import umap
import hdbscan
```

# Load the embedding model

The `all-MiniLM-L6-v2` is a popular, lightweight model designed to convert sentences and paragraphs into dense vector representations (embeddings). It is a staple in the `sentence-transformers` library because it balances speed and performance exceptionally well.

```python
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
```

# Toy Example 1

```python
def generateSimpleToyData():
    """Generates a 1-word dataset of animals and food."""
    animals = ["Lion", "Tiger", "Elephant", "Giraffe", "Zebra", 
               "Monkey", "Kangaroo", "Penguin", "Panda", "Wolf"]
    food = ["Apple", "Banana", "Pizza", "Burger", "Pasta", 
            "Sushi", "Bread", "Cheese", "Chicken", "Salad"]
    
    toy_df = pd.DataFrame({
        "text": animals + food,
        "true_category": (["Animal"] * 10) + (["Food"] * 10)
    })
    return toy_df.sample(frac=1, random_state=42).reset_index(drop=True)

df_simple = generateSimpleToyData()
```

```python
embeddings = embed_model.encode(df_simple['text'].tolist())

# Reduce to 2D for Visualization
reducer = umap.UMAP(n_neighbors=5, min_dist=0.1, n_components=2, metric='cosine', random_state=42)
viz_embeddings = reducer.fit_transform(embeddings)

# Cluster with HDBSCAN
# We set min_cluster_size=2 because of our small dataset
clusterer = hdbscan.HDBSCAN(min_cluster_size=2)
df_simple['cluster_id'] = clusterer.fit_predict(viz_embeddings)
```

```python
# 1. Setup Data
df_toy = generateSyntheticData()

# 2. Embeddings
# Using the same model from your notebook
embed_model = SentenceTransformer('all-MiniLM-L6-v2')
toy_embeddings = embed_model.encode(df_toy['text'].tolist())

# 3. Dimensionality Reduction (UMAP)
# We project to 2D specifically for the visualization step
reducer_2d = umap.UMAP(
    n_neighbors=5, 
    min_dist=0.1, 
    n_components=2, 
    metric='cosine', 
    random_state=42
)
viz_embeddings = reducer_2d.fit_transform(toy_embeddings)

# 4. Clustering (HDBSCAN)
# Using a smaller min_cluster_size for the toy problem
toy_clusterer = hdbscan.HDBSCAN(min_cluster_size=3, min_samples=1)
df_toy['cluster_id'] = toy_clusterer.fit_predict(viz_embeddings)
```

```python
# 5. Visualization
plt.figure(figsize=(10, 7))
sns.scatterplot(
    x=viz_embeddings[:, 0], 
    y=viz_embeddings[:, 1], 
    hue=df_toy['cluster_id'], 
    palette='viridis', 
    s=100
)
plt.title("2D Projection of Synthetic Text Clusters")
plt.show()
```
